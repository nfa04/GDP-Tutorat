## Aufgabentyp 2: Rekursion und Schleife

Ergänze jeweils die fehlende Methode. Sie soll dasselbe Ergebnis wie die vorhandene Methode liefern. Teste auch einen Randfall.

### Fakultät

````java
class Fact {
    public static int facrec(int n) {
        if (n <= 1) {
            return 1;
        }
        return n * facrec(n - 1);
    }

    public static int facloop(int n) {
        int result = 1;
        for (int i = 2; i <= n; i++) {
            result *= i;
        }
        return result;
    }

    public static void main(String[] args) {
        System.out.println(facrec(5));
        System.out.println(facloop(5));
        System.out.println(facloop(0));
    }
}

Fact.main(new String[0]);
````

### Ziffern zählen

````java
class Digits {
    public static int countDigitsRec(int n) {
        if (n >= -9 && n <= 9) {
            return 1;
        }
        return 1 + countDigitsRec(n / 10);
    }

    public static int countDigitsLoop(int n) {
        int count = 1;
        while (n <= -10 || n >= 10) {
            count++;
            n /= 10;
        }
        return count;
    }

    public static void main(String[] args) {
        System.out.println(countDigitsRec(123456));
        System.out.println(countDigitsLoop(123456));
        System.out.println(countDigitsLoop(-42));
    }
}

Digits.main(new String[0]);
````

### Gerade Zahlen

````java
class EvenPrinter {
    public static void printNumbersLoop(int n) {
        for (int i = 1; i <= n; i++) {
            if (i % 2 == 0) {
                System.out.println(i);
            }
        }
    }

    public static void printNumbersRec(int n) {
        if (n <= 0) {
            return;
        }
        printNumbersRec(n - 1);
        if (n % 2 == 0) {
            System.out.println(n);
        }
    }

    public static void main(String[] args) {
        printNumbersRec(7);
    }
}

EvenPrinter.main(new String[0]);
````

### Produkt für `b >= 0`

````java
class Product {
    public static int prodloop(int a, int b) {
        int result = 0;
        while (b > 0) {
            result = result + a;
            b = b - 1;
        }
        return result;
    }

    public static int prodrec(int a, int b) {
        if (b <= 0) {
            return 0;
        }
        return a + prodrec(a, b - 1);
    }

    public static void main(String[] args) {
        System.out.println(prodloop(3, 5));
        System.out.println(prodrec(3, 5));
        System.out.println(prodrec(3, 0));
    }
}

Product.main(new String[0]);
````

### Programm 1

**Programm 1:** Bei `n > 1` endet die Methode sofort; bei `n <= 1` fehlt ein erreichbarer Abbruch. Korrekt ist ein Basisfall `n <= 0`, danach ein Stern und der Aufruf mit `n - 1`.

````java
class Aufgabe1 {
    // Die Methode soll n Sterne ausgeben
    public static void printStars(int n) {
        if (n <= 0) {
            return;
        }
        System.out.print("*");
        printStars(n - 1);
    }

    public static void main(String[] args) {
        printStars(5);
        System.out.println();
    }
}

Aufgabe1.main(new String[0]);
````

### Programm 2

**Programm 2:** `i++` und `i--` heben sich auf. Die Schleife bleibt bei `i = 2`. Nur gerade Werte auszugeben reicht, wenn `i` jeweils um 2 steigt.

````java
class Aufgabe2 {
    // Die Methode soll alle geraden Zahlen zwischen 1 und n ausgeben
    public static void even(int n) {
        for (int i = 2; i <= n; i += 2) {
            System.out.println(i);
        }
    }

    public static void main(String[] args) {
        even(8);
    }
}

Aufgabe2.main(new String[0]);
````

## Aufgabentyp 4: Sternendreieck mit Schleifen

Ergänze `druckeDreieck(int hoehe)` mit zwei verschachtelten `for`-Schleifen. Bei `hoehe = 4` sollen vier Zeilen mit einem, zwei, drei und vier Sternen erscheinen. Bei `hoehe = 0` darf nichts ausgegeben werden.

````java
class SternDreieck {
    public static void druckeDreieck(int hoehe) {
        for (int zeile = 1; zeile <= hoehe; zeile++) {
            for (int spalte = 0; spalte < zeile; spalte++) {
                System.out.print("*");
            }
            System.out.println();
        }
    }

    public static void main(String[] args) {
        druckeDreieck(4);
        druckeDreieck(0);
    }
}

SternDreieck.main(new String[0]);
````

````java
class FantasieProgramm1 {
    public static void test() {
        int a = 5;
        int b = 3;
        int c = a + b;
        System.out.println(c);
        int d = a / b;
        System.out.println(d);
        int e = a % b;
        System.out.println(e);
        double f = (double) a / b;
        System.out.printf(java.util.Locale.US, "%.6f%n", f);
    }

    public static void main(String[] args) {
        test();
    }
}

FantasieProgramm1.main(new String[0]);
````

##### Java-Übersetzung: Programm 2

````java
class FantasieProgramm2 {
    public static int summe(int n) {
        int sum = 0;
        while (n > 0) {
            sum = sum + n;
            n = n - 1;
        }
        return sum;
    }

    public static void main(String[] args) {
        int s = summe(5);
        System.out.println(s);
    }
}

FantasieProgramm2.main(new String[0]);
````

## Aufgabentyp 6: Sierpiński-Dreieck

Schreibe **eine** Zeichenmethode `zeichneSierpinski(double seitenlaenge, int steps)` für `steps >= 0`. Bei `steps = 0` zeichnet sie ein gleichseitiges Dreieck. Sonst zeichnet sie drei kleinere Dreiecke mit halber Seitenlänge: unten links, unten rechts und oben. Nach jedem Aufruf steht die Turtle wieder am Startpunkt und schaut in dieselbe Richtung. In `main` steht nur der Aufruf dieser Methode. Teste `steps = 0`, `1` und `3`.

Wichtige Turtle-Befehle:

- `clear()` leert die Zeichnung; `reset()` setzt die Turtle zurück; `home()` bringt sie zur Mitte.
- `forward(x)` und `backward(x)` bewegen sie um `x` Schritte.
- `turnLeft(w)` und `turnRight(w)` drehen sie um `w` Grad.
- `penUp()` bewegt ohne Linie; `penDown()` zeichnet wieder Linien.
- `display(drawing())` zeigt die fertige Zeichnung an.

````java
import static ch.unibas.informatik.jturtle.TurtleCommands.*;
````

````java
class Sierpinski {
    public static void zeichneSierpinski(double seitenlaenge, int steps) {
        if (steps == 0) {
            for (int i = 0; i < 3; i++) {
                forward(seitenlaenge);
                turnLeft(120);
            }
            return;
        }

        double halb = seitenlaenge / 2;
        zeichneSierpinski(halb, steps - 1);

        penUp();
        forward(halb);
        penDown();
        zeichneSierpinski(halb, steps - 1);

        penUp();
        backward(halb);
        turnLeft(60);
        forward(halb);
        turnRight(60);
        penDown();
        zeichneSierpinski(halb, steps - 1);

        penUp();
        turnLeft(60);
        backward(halb);
        turnRight(60);
        penDown();
    }

    public static void main(String[] args) {
        zeichneSierpinski(80, 3);
    }
}

Sierpinski.main(new String[0]);
display(drawing());
````

Der Basisfall zeichnet ein Dreieck. Jeder weitere Schritt halbiert die Seitenlänge und ruft die Methode dreimal auf. Nach jedem Ortswechsel geht die Turtle ohne Zeichnen zum Ausgangspunkt zurück; daher beginnen alle Aufrufe mit derselben Blickrichtung.
