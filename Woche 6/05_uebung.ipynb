### Ziffern zählen

````java
class Digits {

    public static int countDigitsRec(int n) {
        if (n >= -9 && n <= 9) {
            return 1;
        }
        return 1 + countDigitsRec(n / 10);
    }

    public static int countDigitsLoop(int n) {
        int counter = 0;
        while(n > 10 && n < -10) {
            n = n / 10;
            counter = counter + 1;
        }
        return counter;
    }

    public static void main(String[] args) {
        System.out.println(countDigitsLoop(123456));
    }

}

Digits.main(new String[0]);
````

### Gerade Zahlen

````java
class EvenPrinter {

    public static void printNumbersLoop(int n) {
        for (int i = 1; i <= n; i++) {
            if (i % 2 == 0) {
                System.out.println(i);
            }
        }
    }

    public static void printNumbersRec(int n) {
        int counter = 0;
        while(n > 10 && n < -10) {
            n = n / 10;
            if(n %% 2 == 0) {
                counter = counter + 1;
            }
        }
        return counter;
    }


    public static void main(String[] args) {
        printNumbersLoop(7);
    }

}

EvenPrinter.main(new String[0]);
````
