## 1. Einstieg: Ein Kreis ohne Variablen

Wir wollen für einen Kreis mit Radius 5 Fläche und Umfang ausgeben. Die Formeln sind \(A = \pi r^2\) und \(U = 2\pi r\).

Das folgende Programm rechnet richtig, aber die Bedeutung der Zahlen ist schwer zu erkennen. **Frage an die Gruppe:** An wie vielen Stellen müsste man etwas ändern, wenn der Radius 8 statt 5 wäre?

```java
class KreisOhneVariablen {
    public static void main(String[] args) {
        System.out.println("Fläche: " + 3.141592653589793 * 8 * 8);
        System.out.println("Umfang: " + 2 * 3.141592653589793 * 8);
    }
}
KreisOhneVariablen.main(new String[0]);
```

### Derselbe Kreis mit Variablen

Eine Variable hat einen **Typ**, einen **Namen** und einen **Wert**. In `double radius = 5.0;` wird die Variable `radius` deklariert und gleich initialisiert. Danach verwenden beide Formeln denselben Wert. Für einen anderen Kreis muss nur **eine** Stelle geändert werden.

`Math.PI` liefert die Kreiszahl in Java; die Namen `flaeche` und `umfang` machen die Zwischenergebnisse sichtbar.

```java
class KreisMitVariablen {
    public static void main(String[] args) {
        double radius = 8.0;             // Diese eine Zahl ändern.
        double flaeche = Math.PI * radius * radius;
        double umfang = 2 * Math.PI * radius;

        System.out.println("Radius: " + radius);
        System.out.println("Fläche: " + flaeche);
        System.out.println("Umfang: " + umfang);
    }
}
KreisMitVariablen.main(new String[0]);
```

**Mini-Check:** Ändere `radius` auf `8.0`. Ändern sich Fläche und Umfang wie erwartet? Eine spätere Wertzuweisung sähe etwa so aus: `radius = 8.0;` — ohne erneute Typangabe.

## 3. Typecasting bei Zahlen

**Merksatz:** Eine *Erweiterung* des Zahlenbereichs erlaubt Java normalerweise ohne expliziten Cast, zum Beispiel `int → long` oder `float → double`. Bei einer *Verengung* (z. B. `double → int`) verlangt Java einen expliziten Cast: `(Zieltyp) wert`. Dabei können Nachkommastellen oder Zahlenbereich verloren gehen.

**Genauigkeits-Vorbehalt:** „Automatisch“ bedeutet nicht immer „exakt“. `long → double` ist erlaubt, kann bei sehr grossen Zahlen aber Präzision verlieren. Der Merksatz ist deshalb eine Orientierung, keine Garantie für verlustfreie Speicherung.

Ein Cast wandelt den Wert für den jeweiligen Ausdruck um; die ursprüngliche Variable ändert sich dadurch nicht.

```java
class ZahlenUmwandeln {
    public static void main(String[] args) {
        double preis = 7.9;
        float messwert = 2.5f;
        int anzahl = 42;

        
        // TODO: Wandle preis von double in int um.
        int neuer_preis = (int) preis;
        // TODO: Wandle messwert von float in double um.
        double neuer_messwert = (double) messwert;
        // TODO: Wandle anzahl von int in float um.
        float neue_anzahl = (float) anzahl;
        // TODO: Gib die drei neuen Werte aus.
        System.out.println("preis=" + neuer_preis + " messwert= " + neuer_messwert + " anzahl=" + neue_anzahl);
    }
}
ZahlenUmwandeln.main(new String[0]);
```

## 4. Zwischen `String` und `int` umwandeln; Text ausschneiden

Eine Zahl als Text (`"42"`) ist nicht dasselbe wie die Zahl `42`. Zwischen `String` und `int` benutzen wir Methoden, **keinen** Cast mit `(int)` oder `(String)`.

- `Integer.parseInt("42")`: Text → `int`; der Text muss eine gültige Ganzzahl darstellen.
- `String.valueOf(42)`: `int` → Text.
- `substring(start, ende)`: Teil eines Strings; `start` ist inklusive, `ende` exklusiv. Der erste Index ist `0`.
- `substring(start)`: vom Startindex bis zum Ende.

```java
class TextUndZahl {
    public static void main(String[] args) {
        String textZahl = "42";
        int zahl = Integer.parseInt(textZahl);
        String neuerText = String.valueOf(zahl + 8);

        String woche = "Woche02";
        String wort = woche.substring(0, 5); // Zeichen 0 bis 4: "Woche"
        String nummer = woche.substring(5);  // ab Zeichen 5: "02"

        System.out.println("42 + 8 als Zahl: " + (zahl + 8));
        System.out.println("Zurück als Text: " + neuerText);
        System.out.println(wort + " / " + nummer);
    }
}
TextUndZahl.main(new String[0]);
```

## 5. Division und Modulo

```java
class GanzzahlDivision {
    public static void main(String[] args) {
        System.out.println("7 / 2 = " + (7 / 2));
        System.out.println("2 / 7 = " + (2 / 7));
        System.out.println("-7 / 2 = " + (-7 / 2));
        System.out.println("0 / 7 = " + (0 / 7));
        double ergebnis = 7 / 2;
        System.out.println("double ergebnis = 7 / 2: " + ergebnis);
        // Zum Fehlertest einzeln einkommentieren:
        // System.out.println(7 / 0);
    }
}
GanzzahlDivision.main(new String[0]);
```

```java
class GleitkommaDivision {
    public static void main(String[] args) {
        System.out.println("7 / 2.0 = " + (7 / 2.0));
        System.out.println("7.0 / 2 = " + (7.0 / 2));
        System.out.println("(double) 7 / 2 = " + ((double) 7 / 2));
        System.out.println("(double) (7 / 2) = " + ((double) (7 / 2)));
        System.out.println("7.0 / -2 = " + (7.0 / -2));
        System.out.println("7.0 / 0.0 = " + (7.0 / 0.0));
        System.out.println("0.0 / 0.0 = " + (0.0 / 0.0));
    }
}
GleitkommaDivision.main(new String[0]);
```

```java
class RestDivision {
    public static void main(String[] args) {
        System.out.println("17 % 5 = " + (17 % 5));
        System.out.println("8 % 2 = " + (8 % 2));
        System.out.println("-7 % 2 = " + (-7 % 2));
        System.out.println("7 % -2 = " + (7 % -2));
        System.out.println("2 % 7 = " + (2 % 7));
        System.out.println("0 % 5 = " + (0 % 5));
        // Zum Fehlertest einzeln einkommentieren:
        // System.out.println(5 % 0);
    }
}
RestDivision.main(new String[0]);
```

| Ausdruck | Ergebnis | Warum? |
|---|---:|---|
| `7 / 2` | `3` | `int / int`: Rest fällt weg |
| `2 / 7` | `0` | Ergebnis kleiner als 1 |
| `-7 / 2` | `-3` | Abschneiden Richtung 0 |
| `0 / 7` | `0` | 0 geteilt durch Zahl |
| `double x = 7 / 2` | `3.0` | Division erfolgt vor der Zuweisung |
| `7 / 0` | Fehler | Ganzzahldivision durch 0 |
| `7 / 2.0`, `7.0 / 2` | `3.5` | ein Operand ist `double` |
| `(double) 7 / 2` | `3.5` | Cast vor der Division |
| `(double) (7 / 2)` | `3.0` | Cast nach der Division |
| `7.0 / -2` | `-3.5` | negatives Ergebnis |
| `7.0 / 0.0` | `Infinity` | Gleitkommadivision durch 0 |
| `0.0 / 0.0` | `NaN` | kein definierter Zahlenwert |
| `17 % 5` | `2` | Rest nach 3 vollen Gruppen |
| `8 % 2` | `0` | gerade Zahl |
| `-7 % 2`, `7 % -2` | `-1`, `1` | Rest hat Vorzeichen der linken Zahl |
| `2 % 7` | `2` | Divisor ist grösser |
| `0 % 5` | `0` | Rest von 0 |
| `5 % 0` | Fehler | Restdivision durch 0 |

## 6. Die drei Fehlerarten

| Fehlerart | Wann fällt sie auf? | Auswirkung |
|---|---|---|
| **Übersetzungsfehler** (Kompilierfehler) | vor dem Start | Programm kann so nicht ausgeführt werden |
| **Laufzeitfehler** | während der Ausführung | Programm startet, bricht an der problematischen Stelle ab (wenn nicht behandelt) |
| **Logischer Fehler** | oft erst beim Prüfen der Ergebnisse | Programm läuft, liefert aber das falsche Ergebnis |

```java
class UebersetzungsfehlerDemo {
    public static void main(String[] args) {
        // Fehlendes Semikolon: Zeile für die Demo einkommentieren.
        int zahl = 7
        System.out.println("Übersetzungsfehler");
    }
}
UebersetzungsfehlerDemo.main(new String[0]);
```

```java
class LaufzeitfehlerDemo {
    public static void main(String[] args) {
        int teiler = 0;
        // Division durch 0: Zeile für die Demo einkommentieren.
        System.out.println(5 / teiler);
        System.out.println("Laufzeitfehler");
    }
}
LaufzeitfehlerDemo.main(new String[0]);
```

```java
class LogischerFehlerDemo {
    public static void main(String[] args) {
        double radius = 5.0;
        double flaeche = 2 * Math.PI * radius * radius; // erwartet: Math.PI * radius * radius
        System.out.println("Fläche: " + flaeche);
    }
}
LogischerFehlerDemo.main(new String[0]);
```
