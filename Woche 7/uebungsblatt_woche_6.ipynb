### Aufgabe 1
Das Programm soll die Zahlen im Array ausgeben.


````java
public class Aufgabe1 {
    public static void main(String[] args) {
        int[] zahlen = {10, 20, 30, 40};
        for (int i = 0; i <= zahlen.length; i++) {
            System.out.println(zahlen[i]);
        }
    }
}
````

### Aufgabe 4
Das Programm soll eine Kopie eines Arrays erstellen, und in dieser Kopie die Zahlen verdoppeln.


````java
import java.util.Arrays;

public class Aufgabe4 {
    public static int[] verdoppeln(int[] zahlen) {
        int[] ergebnis = zahlen;
        for (int i = 0; i < ergebnis.length; i++) {
            ergebnis[i] = ergebnis[i] * 2;
        }
        return ergebnis;
    }

    public static void main(String[] args) {
        int[] original = {2, 4, 6};
        int[] neu = verdoppeln(original);
        System.out.println(Arrays.toString(original));
        System.out.println(Arrays.toString(neu));
    }
}
````

### Aufgabe 5
Warum gibt dieses Programm 'Arrays sind verschieden' aus? Behebe den Fehler so, dass auf inhaltliche Gleichheit geprüft wird.


````java
public class Aufgabe5 {
    public static void main(String[] args) {
        int[] a = {1, 2, 3};
        int[] b = {1, 2, 3};
        if (a.equals(b)) {
            System.out.println("Arrays sind gleich");
        } else {
            System.out.println("Arrays sind verschieden");
        }
    }
}
````
