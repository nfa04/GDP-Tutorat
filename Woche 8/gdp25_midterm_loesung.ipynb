## Aufgabe 1 (2 Punkte)

Der folgende Code soll eine Gleitkommazahl auf die erste Kommastelle runden.

```java
class CommaRound {
    public static double commaRound(double x) {
        x = x * 10;
        int rounded = Math.round(x);
        x = rounded / 10.0;
        return x;
    }

    public static void main(String[] args) {
        double x = 3.1415;
        double rounded = commaRound(x);
        System.out.println(rounded)
    }
}
```

Der Code kompiliert jedoch nicht. Finden Sie die Fehler und fügen an der entsprechenden Stelle im Code einen Kommentar hinzu, der den Fehler beschreibt. Korrigieren Sie zudem das Programm, so dass es korrekt funktioniert.

````java
class CommaRound {
    public static double commaRound(double x) {
        x = x * 10;
        // Fehler 1: Math.round(double) liefert einen long. Ein long passt nicht ohne
        // Weiteres in einen int (möglicher Informationsverlust) -> expliziter Cast nötig.
        int rounded = (int) Math.round(x);
        x = rounded / 10.0;
        return x;
    }

    public static void main(String[] args) {
        double x = 3.1415;
        double rounded = commaRound(x);
        // Fehler 2: Am Ende der Anweisung fehlte das Semikolon.
        System.out.println(rounded);
    }
}
````

````java
CommaRound.main(new String[0]);
````

**Erklärung:** Es gibt zwei Fehler:
1. `Math.round(double)` gibt einen `long` zurück. Die Zuweisung `long -> int` ist ohne Cast ein Compilerfehler (*incompatible types: possible lossy conversion from long to int*). Lösung: `(int) Math.round(x)`. Alternativ könnte man `rounded` als `long` deklarieren.
2. Das Semikolon nach `System.out.println(rounded)` fehlt.

Die Division `rounded / 10.0` ist dagegen schon richtig: Durch `10.0` (ein `double`) findet eine Gleitkommadivision statt, `31 / 10.0 = 3.1`. Mit `/ 10` würde dagegen eine Ganzzahldivision stattfinden (`31 / 10 = 3`).

Erwartete Ausgabe: `3.1`

## Aufgabe 2 (2 Punkte)

Der folgende Code soll den Stundenlohn einer Person berechnen, basierend auf einem Tagesgehalt und der Anzahl der Arbeitsstunden pro Tag. Das Resultat ist aber falsch. Korrigieren Sie den Code so, dass das Resultat richtig wird.

```java
class SalaryCalculator {
    public static void main(String[] args) {
        int dailySalary = 250;
        int workingHoursPerDay = 8;
        double hourlyRate = dailySalary / workingHoursPerDay;
        System.out.println("Der Stundenlohn beträgt: " + hourlyRate + " Franken.");
    }
}
```

````java
class SalaryCalculator {
    public static void main(String[] args) {
        int dailySalary = 250;
        int workingHoursPerDay = 8;
        // 250 / 8 wäre eine Ganzzahldivision (= 31). Mit 8.0 bzw. einem Cast
        // auf double wird als Gleitkommazahl gerechnet: 250 / 8.0 = 31.25
        double hourlyRate = (double) dailySalary / workingHoursPerDay;
        System.out.println("Der Stundenlohn beträgt: " + hourlyRate + " Franken.");
    }
}
````

````java
SalaryCalculator.main(new String[0]);
````

**Erklärung:** Sind beide Operanden von `/` vom Typ `int`, ist auch das Resultat eine Ganzzahl: `250 / 8 = 31`, der Nachkommateil wird abgeschnitten. Erst danach wird `31` in `31.0` umgewandelt und `hourlyRate` zugewiesen. Man muss also **vor** der Division einen Operanden zu `double` machen:
* `(double) dailySalary / workingHoursPerDay` (der Cast bindet stärker als `/`, er gilt nur für `dailySalary`)
* oder `dailySalary / (double) workingHoursPerDay`
* oder `dailySalary / 8.0`.

Falsch wäre `(double) (dailySalary / workingHoursPerDay)`: Hier passiert der Cast erst nach der Ganzzahldivision, das Resultat ist wieder `31.0`.

Erwartete Ausgabe: `Der Stundenlohn beträgt: 31.25 Franken.`

## Aufgabe 3 (4 Punkte)
Schreiben Sie eine Methode `twinPrimes` welche zwei ganzzahlige Parameter `lo` und `hi` entgegennimmt und alle ganzen Zahlen $n$ zwischen `lo` und `hi` (Grenzen inklusive) ausgibt, für die sowohl $n$ als auch $n+2$ prim sind. (Für uns ist 1 keine Primzahl). Beim Aufruf `twinPrimes(2, 11)` müsste die Methode also die Zahlen 3, 5 und 11 ausgeben.

````java
class Numbers {

    // Hilfsmethode: Ist n eine Primzahl? (1 und kleiner zählen nicht als Primzahl)
    public static boolean isPrime(int n) {
        if (n <= 1) {
            return false;
        }
        // Es genügt, Teiler bis zur Wurzel von n zu testen: i * i <= n
        for (int i = 2; i * i <= n; i++) {
            if (n % i == 0) {
                return false;       // i ist ein Teiler -> keine Primzahl
            }
        }
        return true;
    }

    public static void twinPrimes(int lo, int hi) {
        for (int n = lo; n <= hi; n++) {
            if (isPrime(n) && isPrime(n + 2)) {
                System.out.println(n);
            }
        }
    }

    public static void main(String[] args) {
        // Aufruf für lo = 2 und hi = 11
        twinPrimes(2, 11);
    }

}
````

````java
Numbers.main(new String[0]);
````

**Erklärung:** Die Aufgabe zerfällt in zwei Teile: Zuerst schreiben wir eine Hilfsmethode `isPrime`, dann benutzen wir sie in einer Schleife.

* `isPrime(n)`: `n <= 1` ist nie prim. Sonst probieren wir alle möglichen Teiler `i` ab 2. Findet sich einer (`n % i == 0`), ist `n` nicht prim. Man muss nur bis `i * i <= n` testen, denn hat `n` einen Teiler grösser als `√n`, dann hat es auch einen kleineren. (Alle `i` bis `n - 1` zu testen wäre aber auch korrekt.)
* `twinPrimes`: Für jedes `n` von `lo` bis `hi` (beide inklusive, deshalb `<=`) testen wir, ob `n` **und** `n + 2` prim sind.

Beachten Sie, dass `n + 2` auch über `hi` hinausgehen darf: Bei `n = 11` ist `13` zu prüfen, obwohl `13 > hi`. Nur `n` selbst muss im Bereich liegen.

Erwartete Ausgabe für `twinPrimes(2, 11)`: `3`, `5`, `11` (Paare: (3,5), (5,7), (11,13); (2,4) und (7,9) sind keine).

## Aufgabe 4 (4 Punkte)

Schreiben Sie eine Methode `makeGap`, welche zwei Strings und eine ganze Zahl `gap` entgegennimmt. Die Methode soll einen String zurückgeben, in dem die beiden Eingabestrings mit `gap` vielen Leerzeichen dazwischen stehen. So soll zum Beispiel `makeGap("Java", "Fun", 1)` den String `Java Fun` zurückgeben. Sie dürfen, müssen aber nicht `String.format` verwenden. Falls Sie `String.format` verwenden können Sie das Formatierungszeichen `%s` für Strings verwenden. 

````java
class Gap {

    public static String makeGap(String s1, String s2, int gap) {
        String result = s1;
        for (int i = 0; i < gap; i++) {
            result = result + " ";      // gap-mal ein Leerzeichen anhängen
        }
        result = result + s2;
        return result;

        // Alternativen:
        // return s1 + " ".repeat(gap) + s2;
        // return String.format("%s%s%s", s1, " ".repeat(gap), s2);
    }

    public static void main(String[] args) {
        System.out.println(makeGap("Java", "Fun", 1));
        System.out.println(makeGap("Java", "Fun", 2));
    }
}
````

````java
Gap.main(new String[0]);
````

**Erklärung:** Wir starten mit dem ersten String, hängen in einer Schleife `gap`-mal ein Leerzeichen an und hängen am Schluss den zweiten String an. Mit `+` werden Strings aneinandergehängt.

Die Methode **gibt** den String **zurück** (`return`), die Ausgabe passiert erst in `main`. Wer stattdessen in `makeGap` `System.out.println` verwendet, hat keinen Rückgabewert und verliert Punkte.

Mit `" ".repeat(gap)` (seit Java 11) geht es kürzer. Die Schleife funktioniert aber überall und ist völlig in Ordnung.

## Aufgabe 5 (4 Punkte)
Schreiben Sie eine Methode `drawPyramid` welche eine positive ganze Zahl `n` entgegennimmt und eine Pyramide von Höhe `n` ausgibt. Eine Pyramide besteht hier aus Sternen `*`, die so angeordnet sind, dass in jeder höheren Schicht sowohl links als auch recht ein Stern weniger steht. Für $n=4$ sollten Sie die folgende Pyramide erhalten:

```
   *   
  ***  
 ***** 
*******
```

````java
class Pyramid {

    public static void drawPyramid(int n) {
        int width = 2 * n - 1;                       // Breite der untersten Zeile
        int middle = n - 1;                          // Spalte der Spitze

        for (int row = 0; row < n; row++) {          // Zeile 0 = Spitze
            for (int col = 0; col < width; col++) {
                // In Zeile 'row' stehen Sterne bis zu 'row' Spalten links/rechts der Mitte
                if (col >= middle - row && col <= middle + row) {
                    System.out.print("*");
                } else {
                    System.out.print(" ");
                }
            }
            System.out.println();                    // Zeilenumbruch
        }
    }

    public static void main(String[] args) {
        drawPyramid(4);
    }

}
````

````java
Pyramid.main(new String[0]);
````

**Erklärung:** Die Pyramide für `n = 4` ist `2n - 1 = 7` Zeichen breit, die Spitze steht in der Mitte (Spalte `n - 1 = 3`, Zählung ab 0). Zeile `row` (0 = oberste) enthält `2 * row + 1` Sterne: `1, 3, 5, 7`. Sie reichen von Spalte `middle - row` bis `middle + row`. Alle anderen Spalten sind Leerzeichen.

```
Zeile 0:    *      Spalten 3..3
Zeile 1:   ***     Spalten 2..4
Zeile 2:  *****    Spalten 1..5
Zeile 3: *******   Spalten 0..6
```

Wie bei jedem Muster aus Zeilen und Spalten: äussere Schleife = Zeilen, innere Schleife = Spalten, am Ende jeder Zeile ein `println()`.

**Variante:** Alternativ kann man pro Zeile zuerst `n - 1 - row` Leerzeichen und dann `2 * row + 1` Sterne drucken. Die rechten Leerzeichen sind unsichtbar, deshalb ist es in Ordnung, sie wegzulassen.

## Aufgabe 6 (3 Punkte)
Schreiben Sie eine rekursive Methode `sumOfSquares` welche eine ganze Zahl `n` entgegennimmt und die Summe aller Quadrate der Zahlen von $1$ bis `n` zurückgibt, also $1^2 + 2^2 + \ldots + n^2$. Für $n=5$ sollten Sie das Resultat $55$ erhalten.

````java
class Recursion {

    public static int sumOfSquares(int n) {
        if (n <= 0) {                           // Basisfall: leere Summe
            return 0;
        }
        return n * n + sumOfSquares(n - 1);     // Rekursionsschritt
    }

    public static void main(String[] args) {
        System.out.println(sumOfSquares(5));
    }
}
````

````java
Recursion.main(new String[0]);
````

**Erklärung:** Eine rekursive Methode braucht einen **Basisfall** (hier: die leere Summe `n = 0` ist `0`) und einen **Rekursionsschritt**, der das Problem auf ein kleineres zurückführt:

$$1^2 + 2^2 + \ldots + n^2 = n^2 + \left(1^2 + \ldots + (n-1)^2\right)$$

also `sumOfSquares(n) = n * n + sumOfSquares(n - 1)`.

Ablauf für `n = 3`: `3*3 + sumOfSquares(2)` = `9 + (4 + (1 + 0))` = `14`.

Wir prüfen `n <= 0` statt `n == 0`, damit die Methode bei negativen Eingaben nicht endlos rekursiert (`StackOverflowError`).

Erwartetes Resultat für `n = 5`: `1 + 4 + 9 + 16 + 25 = 55`

## Aufgabe 7 (3 Punkte)

Schreiben Sie eine Methode `assignments`, die zwei ganze Zahlen entgegen nimmt: die Anzahl Studierenden und die maximale Grösse einer Übungsgruppe. Die Methode soll berechnen, wieviele Übungsgruppen mindestens erstellt werden müssen, damit alle Studierenden in einer Gruppe unterkommen und keine Gruppe zu gross ist. Die Methode soll die Lösung in einem Satz der Form `Es braucht mindestens X Übungsgruppen` ausgeben, wobei `X` durch den korrekten Wert ersetzt werden soll. Zudem soll `X` als ganze Zahl angezeigt werden, also z.B. als `2`, und nicht als `2.0`.

````java
class Groups {

    public static void assignments(int students, int maxSize) {
        // Aufrunden von students / maxSize mit reiner Ganzzahlarithmetik
        int groups = (students + maxSize - 1) / maxSize;
        System.out.println("Es braucht mindestens " + groups + " Übungsgruppen");

        // Alternative mit Fallunterscheidung:
        // int groups = students / maxSize;
        // if (students % maxSize != 0) { groups = groups + 1; }
    }

    public static void main(String[] args) {
        // Berechnung für 135 Studierende mit Gruppengrösse 7
        assignments(135, 7);
    }
}
````

````java
Groups.main(new String[0]);
````

**Erklärung:** Gesucht ist `students / maxSize`, **aufgerundet**. `135 / 7 = 19.28…`, also braucht es 20 Gruppen (19 volle Gruppen mit 133 Studierenden, und die restlichen 2 brauchen eine weitere Gruppe).

Die Ganzzahldivision rundet ab. Zum Aufrunden gibt es zwei gängige Wege:
* **Fallunterscheidung:** `students / maxSize` rechnen und 1 addieren, falls ein Rest bleibt (`students % maxSize != 0`).
* **Trick:** `(students + maxSize - 1) / maxSize`. Durch das Addieren von `maxSize - 1` wird jeder Rest zu einer vollen zusätzlichen Gruppe, aber ohne Rest ändert sich nichts.

Man könnte auch `(int) Math.ceil((double) students / maxSize)` schreiben. Hier ist der `double`-Cast wichtig, sonst wird schon vor `Math.ceil` abgerundet. Und wegen der Vorgabe "als `2` und nicht `2.0`" braucht es dann den Cast auf `int` für die Ausgabe.

Erwartete Ausgabe: `Es braucht mindestens 20 Übungsgruppen`

## Aufgabe 8 (3 Punkte)
Schreiben Sie eine Methode `add` welche ein Gleitkommazahl-Arrays und eine Gleitkommazahl `x` entgegennimmt und ein Array zurückgibt, dessen $i$'ter Eintrag die Summe des jeweiligen Eintrags mit `x` ist. Der Aufruf `add({1.2, 2.5, 1.0}, 2.0)` sollte also das Array `{3.2, 4.5, 3.0}` zurückgeben.

````java
import java.util.Arrays;

class ArraySum {

    public static double[] add(double[] arr, double x) {
        double[] result = new double[arr.length];   // neues Array gleicher Länge
        for (int i = 0; i < arr.length; i++) {
            result[i] = arr[i] + x;
        }
        return result;
    }

    public static void main(String[] args) {
        System.out.println(Arrays.toString(add(new double[]{1.2, 2.5, 1.0}, 2.0)));
    }
}
````

````java
ArraySum.main(new String[0]);
````

**Erklärung:** Wir legen ein neues Array derselben Länge an, gehen alle Indizes durch und speichern `arr[i] + x` an derselben Stelle. Das ursprüngliche Array wird nicht verändert. (Hätte man `arr[i] = arr[i] + x` geschrieben und `arr` zurückgegeben, wäre das Eingabe-Array verändert worden. Die Aufgabe sagt aber, dass ein Array *zurückgegeben* werden soll.)

*Gleitkommazahl* ist der Oberbegriff für `float` und `double`. Da `main` `double`-Werte übergibt, ist der passende Typ `double[]`.

Erwartete Ausgabe: `[3.2, 4.5, 3.0]`

## Aufgabe 9 (5 Punkte)
Für diese Aufgabe geben wir Ihnen bereits eine Methode, die Sie weder verstehen noch ändern müssen. Die Methode `cross(x1, y1, x2, y2, x3, y3, x4, y4)` schaut sich die vier Punkte $p_1 = (x_1, y_1)$ ,$p_2 = (x_2, y_2)$, $p_3 = (x_3, y_3)$ und $p_4 = (x_4, y_4)$ in der Ebene an und gibt zurück, ob sich die zwei Segmente $p_1p_2$ und $p_3p_4$ schneiden. Wenn ja, gibt die Methode `true` zurück, ansonsten `false`.

In der Methode `main` sind ausserdem 7 Punkte $p_1,\ldots,p_7$ als Arrays von zwei Gleitkommazahlen definiert. Ihre Aufgabe ist die folgende:
* Schreiben Sie eine Methode `thrackle`, welche ein Array von Punkten (gespeichert als Arrays von zwei Gleitkommazahlen) entgegennimmt und `true` zurückgibt, wenn vier aufeinanderfolgende Punkte $p_1, p_2, p_3, p_4$ im Array immer zwei kreuzende Segmente $p_1p_2$ und $p_3p_4$ bilden. Wenn es vier aufeinanderfolgende Punkte gibt, die keine solchen Kreuzenden Segmente bilden, soll die Methode `false` zurückgeben.
* Testen Sie Ihre Methode, indem Sie die gegebenen 7 Punkte ins korrekte Format bringen, ihrer Methode übergeben, und den Rückgabewert ausgeben.

````java
class Turns {

    // Diese Methode sollen Sie nicht verändern
    public static boolean cross(double x1, double y1, double x2, double y2, double x3, double y3, double x4, double y4) {
        if (Math.signum(((y2 - y1) * (x3 - x2)) - ((x2 - x1) * (y3 - y2))) != Math.signum(((y2 - y1) * (x4 - x2)) - ((x2 - x1) * (y4 - y2))) &&
            Math.signum(((y4 - y3) * (x2 - x4)) - ((x4 - x3) * (y2 - y4))) != Math.signum(((y4 - y3) * (x1 - x4)) - ((x4 - x3) * (y1 - y4)))) {
            return true;
        }
        return false;
    }

    // Ihre Lösung
    public static boolean thrackle(double[][] points) {
        // Alle Viererblöcke (i, i+1, i+2, i+3) aufeinanderfolgender Punkte
        for (int i = 0; i + 3 < points.length; i++) {
            double[] p1 = points[i];
            double[] p2 = points[i + 1];
            double[] p3 = points[i + 2];
            double[] p4 = points[i + 3];

            // Segmente p1p2 und p3p4 müssen sich kreuzen
            if (!cross(p1[0], p1[1], p2[0], p2[1], p3[0], p3[1], p4[0], p4[1])) {
                return false;           // ein einziger Viererblock ohne Kreuzung genügt
            }
        }
        return true;                    // alle Viererblöcke kreuzen sich
    }

    public static void main(String[] args) {
        double[] p1 = new double[]{-1.72, -0.18};
        double[] p2 = new double[]{1.84, 3.68};
        double[] p3 = new double[]{-2.38, 3.98};
        double[] p4 = new double[]{1.66, -0.48};
        double[] p5 = new double[]{0.08, 5.38};
        double[] p6 = new double[]{-0.32, 0.36};
        double[] p7 = new double[]{3.44, 2.74};

        // Die 7 Punkte ins richtige Format bringen: ein Array von Punkten (double[][])
        double[][] points = {p1, p2, p3, p4, p5, p6, p7};

        System.out.println(thrackle(points));
    }
}
````

````java
Turns.main(new String[0]);
````

**Erklärung:** Ein Punkt ist ein `double[]` der Länge 2 (`{x, y}`), mehrere Punkte sind ein `double[][]`. Die Aufgabe hat zwei Teile:

1. **Format:** Die sieben einzelnen Punkte werden in ein Array `{p1, ..., p7}` gepackt.
2. **`thrackle`:** Wir gehen alle Viererblöcke `(points[i], points[i+1], points[i+2], points[i+3])` durch und fragen die **gegebene** Methode `cross` mit den Koordinaten (`p[0]` = x, `p[1]` = y), ob sich die Segmente `p1p2` und `p3p4` kreuzen.
   * Sobald ein Block **nicht** kreuzt, ist die Antwort `false`, wir können sofort abbrechen.
   * Nur wenn alle Blöcke kreuzen, ist die Antwort `true`.

**Schleifengrenze:** Der letzte Viererblock beginnt bei `i = length - 4`. Die Bedingung `i + 3 < points.length` stellt sicher, dass `points[i + 3]` noch existiert. Sonst gibt es eine `ArrayIndexOutOfBoundsException`. Bei 7 Punkten gibt es `7 - 3 = 4` Blöcke: `(p1..p4)`, `(p2..p5)`, `(p3..p6)`, `(p4..p7)`.

Bei weniger als 4 Punkten gibt es keinen Block, und die Methode gibt (leer erfüllt) `true` zurück.

Erwartete Ausgabe für die gegebenen Punkte: `true`
