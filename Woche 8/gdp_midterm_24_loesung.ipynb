## Aufgabe 1 (2 Punkte)

Der folgende Code soll den Wurf einer Münze simulieren, wobei `1`für "Kopf" und `0`für Zahl steht.

```java
class Coin {
    public static void main(String[] args) {
        double random_number = Math.random();
        int result = Math.round(random_number);
        System.out.println(result)
    }
}
```

Der Code kompiliert jedoch nicht. Finden Sie die Fehler und fügen an der entsprechenden Stelle im Code einen Kommentar hinzu, der den Fehler beschreibt. Korrigieren Sie zudem das Programm, so dass es korrekt funktioniert.

````java
class Coin {
    public static void main(String[] args) {
        double random_number = Math.random();
        // Fehler 1: Math.round(double) liefert einen long. Ein long passt nicht
        // einfach so in einen int (möglicher Informationsverlust) -> expliziter Cast nötig.
        int result = (int) Math.round(random_number);
        // Fehler 2: Am Ende der Anweisung fehlte das Semikolon.
        System.out.println(result);
    }
}
````

````java
Coin.main(new String[0]);
````

**Erklärung:** Es gibt zwei Fehler: Das fehlende Semikolon nach `println(result)` und die Typumwandlung. `Math.round(double)` gibt einen `long` zurück, und die Zuweisung `long -> int` ist ohne Cast ein Compilerfehler (*incompatible types: possible lossy conversion*). Statt dem Cast hätte man auch `long result` verwenden können, aber die Aufgabe nennt `1`/`0` als Ganzzahl, und `int` ist der naheliegende Typ.

## Aufgabe 2 (2 Punkte)

Der folgende Code soll den Durchschnitt zweier ganzen Zahlen berechnen. Das Resultat ist aber falsch. Korrigieren Sie den Code so, dass das Resultat richtig wird.

```java
class Average {
    public static void main(String[] args) {
        int a = 6;
        int b = 1;
        double average = (a + b) / 2;
        System.out.println(average);
    }

```

````java
class Average {
    public static void main(String[] args) {
        int a = 6;
        int b = 1;
        // (a + b) / 2 ist eine Ganzzahldivision (7 / 2 = 3), erst danach wird zu double.
        // Mit 2.0 (double) wird die Division als Gleitkommadivision ausgeführt: 7 / 2.0 = 3.5
        double average = (a + b) / 2.0;
        System.out.println(average);
    }
}
````

````java
Average.main(new String[0]);
````

**Erklärung:** Sind beide Operanden von `/` ganzzahlig, wird auch das Resultat ganzzahlig (der Nachkommateil wird abgeschnitten): `7 / 2 = 3`. Erst danach würde `3` zu `3.0` umgewandelt. Man muss also **vor** der Division einen Operanden zu `double` machen, z.B. mit `2.0`, oder `(double) (a + b) / 2`. Die Variante `(double) ((a + b) / 2)` wäre hingegen *falsch*, da der Cast erst nach der Ganzzahldivision passiert.

Erwartetes Resultat: `3.5`

## Aufgabe 3 (4 Punkte)
Schreiben Sie eine Methode `nonSquareDivisible` welche zwei ganzzahlige Parameter `n` und `k` entgegennimmt und alle ganzen Zahlen zwischen 1 und $n$ ausgibt, die durch $k$ teilbar sind, aber keine Quadratzahl sind.

````java
class Numbers {

    public static void nonSquareDivisible(int n, int k) {
        for (int i = 1; i <= n; i++) {
            // Quadratzahl? -> Wurzel runden und prüfen, ob r * r wieder i ergibt
            int r = (int) Math.round(Math.sqrt(i));
            boolean isSquare = (r * r == i);

            if (i % k == 0 && !isSquare) {
                System.out.println(i);
            }
        }
    }

    public static void main(String[] args) {
        // Aufruf für n = 16 und k = 3
        nonSquareDivisible(16, 3);
    }

}
````

````java
Numbers.main(new String[0]);
````

**Erklärung:** Wir gehen alle `i` von 1 bis `n` durch und geben `i` aus, wenn
1. `i % k == 0` (teilbar durch `k`) **und**
2. `i` keine Quadratzahl ist.

Eine Zahl `i` ist eine Quadratzahl, wenn es ein ganzes `r` mit `r * r == i` gibt. Wir berechnen dazu `r` als gerundete Wurzel von `i` und prüfen `r * r == i`.

Achtung: Die Bedingung `k * k != i` testet nur, ob `i` das Quadrat von **`k`** ist. Das ist nicht dasselbe wie "`i` ist irgendeine Quadratzahl". Mit `k = 2` würde zum Beispiel `16` fälschlicherweise ausgegeben (`2 * 2 != 16`), obwohl `16 = 4 * 4` eine Quadratzahl ist.

Erwartete Ausgabe für `n = 16`, `k = 3`: `3, 6, 12, 15` (die 9 ist durch 3 teilbar, aber eine Quadratzahl, daher fehlt sie).

````java
Fraction.main(new String[0]);
````

## Aufgabe 5 (4 Punkte)
Schreiben Sie eine Methode `drawIdentity` welche eine positive ganze Zahl `n` entgegennimmt und die Identitätsmatrix von Grösse $n$ ausgibt. (die Identitätsmatrix eine Tabelle mit $n$ Zeilen und $n$ Spalten, wobei fast überall eine 0 steht, ausser auf der Diagonale, d.h. da wo die Zeilennummer und die Spaltennummer gleich sind. Dort steht eine 1.) Für $n=4$ sollten Sie die folgende Matrix erhalten:

```
1000
0100
0010
0001
```

````java
class Matrix {

    public static void drawIdentity(int n) {
        for (int row = 0; row < n; row++) {          // Zeilen
            for (int col = 0; col < n; col++) {      // Spalten
                if (row == col) {
                    System.out.print("1");           // Diagonale
                } else {
                    System.out.print("0");           // sonst
                }
            }
            System.out.println();                    // Zeilenumbruch nach jeder Zeile
        }
    }

    public static void main(String[] args) {
        drawIdentity(4);
    }

}
````

````java
Matrix.main(new String[0]);
````

**Erklärung:** Zwei geschachtelte Schleifen: die äussere läuft über die Zeilen, die innere über die Spalten. In jeder Zelle wird `1` gedruckt, falls Zeilen- und Spaltennummer gleich sind, sonst `0`. Nach jeder Zeile (also nach der inneren Schleife) folgt ein Zeilenumbruch.

**Typische Fehler:**
* Beide Schleifen müssen **genau `n`-mal** laufen. Startet man bei `1`, muss die Bedingung `<= n` heissen, startet man bei `0`, dann `< n`. Mixt man das (`i` von 1 bis `<= n`, `j` von 1 bis `< n`), hat man in jeder Zeile eine Spalte zu wenig. Genau deshalb fehlt dann die letzte `1`.
* Die `0` darf nur im `else`-Fall gedruckt werden. Ohne `else` wird nach jeder `1` zusätzlich eine `0` gedruckt, und die Zeilen werden zu lang.

## Aufgabe 6 (3 Punkte)
Schreiben Sie eine Methode `fibonacci` welche die $n$'te Fibonacci-Zahl $f_n$ zurückgibt. Die Fibonacci-Zahlen sind wie folgt definiert:
* $f_0 = 1$
* $f_1 = 1$
* $f_n = f_{n-1} + f_{n-2}$.

````java
class Fibonacci {

    // Rekursive Lösung (direkt nach der Definition)
    public static int fibonacci(int n) {
        if (n <= 1) {                       // f_0 = 1 und f_1 = 1
            return 1;
        }
        return fibonacci(n - 1) + fibonacci(n - 2);
    }

    // Iterative Lösung (viel schneller für grosse n)
    public static int fibonacciIterative(int n) {
        int previous = 1;                   // f_(i-1)
        int current = 1;                    // f_i
        for (int i = 2; i <= n; i++) {
            int next = previous + current;
            previous = current;
            current = next;
        }
        return current;
    }

    public static void main(String[] args) {
        System.out.println(fibonacci(6));            // 13
        System.out.println(fibonacciIterative(6));   // 13
    }
}
````

````java
Fibonacci.main(new String[0]);
````

**Erklärung:** Die Definition lässt sich fast wörtlich in eine rekursive Methode übersetzen. Wichtig sind **beide** Basisfälle `f_0 = 1` und `f_1 = 1`. Fehlt `f_1`, ruft sich `fibonacci(1)` mit `fibonacci(-1)` auf und läuft endlos (`StackOverflowError`).

Folge: 1, 1, 2, 3, 5, 8, 13, … -> `f_6 = 13`.

Die rekursive Version berechnet dieselben Werte sehr oft neu, was für grosse `n` langsam wird. Die iterative Version merkt sich nur die letzten zwei Werte und braucht nur `n` Schritte.

**Hinweis zur Versuchsversion:** `n * fibonacci(n - 1)` ist die Fakultät und nicht Fibonacci.

## Aufgabe 7 (3 Punkte)

Schreiben Sie eine Methode `assignments`, die zwei ganze Zahlen entgegen nimmt: die Anzahl Studierenden und die Anzahl Übungsgruppen. Die Methode soll berechnen, wieviele Studierende bei einer möglichst gleichmässigen Gruppenaufteilung in einer Übungsgruppe sind und soll die Lösung in einem Satz der Form `Es gibt A Übungsgruppen mit X Studierenden und B Übungsgruppen mit Y Studierenden` ausgeben (wobei `A`, `B`, `X` und `Y` durch die korrekten Werte ersetzt werden sollen).

````java
class Groups {

    public static void assignments(int students, int groups) {
        int small = students / groups;       // Grösse der kleineren Gruppen (abgerundet)
        int rest = students % groups;        // so viele Gruppen haben einen Studierenden mehr

        int bigGroups = rest;                // Gruppen mit small + 1 Studierenden
        int smallGroups = groups - rest;     // Gruppen mit small Studierenden

        if (rest == 0) {
            // Alle Gruppen gleich gross
            System.out.println("Es gibt " + groups + " Übungsgruppen mit " + small + " Studierenden");
        } else {
            System.out.println("Es gibt " + smallGroups + " Übungsgruppen mit " + small
                    + " Studierenden und " + bigGroups + " Übungsgruppen mit " + (small + 1) + " Studierenden");
        }
    }

    public static void main(String[] args) {
        // Berechnung für 135 Studierende und 7 Übungsgruppen
        assignments(135, 7);
        assignments(133, 7);
    }
}
````

````java
Groups.main(new String[0]);
````

**Erklärung:** Mit der Ganzzahldivision `students / groups` erhält man die Grundgrösse `q` jeder Gruppe, mit dem Modulo `students % groups` den Rest `r`. Bei einer möglichst gleichmässigen Aufteilung bekommen genau `r` Gruppen **je einen** zusätzlichen Studierenden:
* `r` Gruppen mit `q + 1` Studierenden
* `groups - r` Gruppen mit `q` Studierenden

Beispiel: `135 = 7 * 19 + 2` -> 5 Gruppen mit 19 und 2 Gruppen mit 20 Studierenden (Kontrolle: `5 * 19 + 2 * 20 = 135`).

**Typischer Fehler:** den ganzen Rest in *eine* Gruppe zu stecken (`q + r`). Das ist nicht gleichmässig: Bei 135 und 7 gäbe das 6 Gruppen mit 19 und eine mit 21 Studierenden. Gleichmässig heisst, dass sich die Gruppengrössen um höchstens 1 unterscheiden.

Für `r == 0` (alle gleich gross) geben wir eine Variante des Satzes ohne den zweiten Teil aus. Das ist nicht streng verlangt, aber ein Satz mit "0 Übungsgruppen" liest sich seltsam.

## Aufgabe 8 (4 Punkte)
Schreiben Sie eine Methode 'sumUp' welche zwei Gleitkommazahl-Arrays von gleicher Länge entgegennimmt und ein Array zurückgibt, dessen $i$'ter Eintrag die Summe der jeweiligen Einträge der entgegengenommenen Arrays ist.

````java
import java.util.Arrays;

class ArraySum {

    public static double[] sumUp(double[] arr1, double[] arr2) {
        double[] result = new double[arr1.length];    // gleich lang wie die Eingaben
        for (int i = 0; i < arr1.length; i++) {
            result[i] = arr1[i] + arr2[i];
        }
        return result;
    }

    public static void main(String[] args) {
        System.out.println(Arrays.toString(sumUp(new double[]{1.2, 2.5, 1.0}, new double[]{2.1, 2.3, 3.0})));
    }
}
````

````java
ArraySum.main(new String[0]);
````

**Erklärung:** Wir legen ein neues Array derselben Länge an, summieren in einer Schleife Eintrag für Eintrag und geben es zurück. Die Eingabe-Arrays werden nicht verändert.

**Zu "Gleitkommazahl":** Gleitkommazahl (*floating point number*) ist der Oberbegriff und umfasst sowohl `float` als auch `double`. Da in der `main`-Methode `double`-Arrays übergeben werden, ist der passende Typ hier `double[]`. Wäre `float[]` gemeint, müsste `main` Literale wie `1.2f` verwenden.

Die Längengleichheit wird laut Aufgabe vorausgesetzt, ein Check ist deshalb nicht nötig. Würde man ihn einbauen, wäre es besser, eine Exception zu werfen (`throw new IllegalArgumentException(...)`), als einfach `arr1` zurückzugeben, da dies stillschweigend ein falsches Resultat liefert.

Erwartete Ausgabe: `[3.3, 4.8, 4.0]` (evtl. mit kleinen Rundungsfehlern wie `3.3000000000000003`, das ist bei `double` normal).

## Aufgabe 9 (5 Punkte)
Für diese Aufgabe geben wir Ihnen bereits eine Methode, die Sie weder verstehen noch ändern müssen. Die Methode `rightTurn(x1, y1, x2, y2, x3, y3)` schaut sich die drei Punkte $p_1 = (x_1, y_1)$ ,$p_2 = (x_2, y_2)$ und $p_3 = (x_3, y_3)$ in der Ebene an und gibt zurück ob man, wenn man bei $p_1$ startet und zu $p_2$ und dann zu $p_3$ geht, eine Rechtskurve gemacht hat. In anderen Worten: die Methode entscheidet, ob $p_3$ auf der rechten Seite der orientierten Gerade $\overrightarrow{p_1p_2}$ liegt. Wenn ja, gibt die Methode `true` zurück, ansonsten `false`.

In der Methode `main` sind ausserdem 7 Punkte $p_1,\ldots,p_7$ als Arrays von zwei Gleitkommazahlen definiert. Ihre Aufgabe ist die folgende:
* Schreiben Sie eine Methode, welche ein Array von Punkten (gespeichert als Arrays von zwei Gleitkommazahlen) entgegennimmt und `true` zurückgibt, wenn drei aufeinanderfolgende Punkte im Array immer eine Rechtskurve bilden. Wenn es drei aufeinanderfolgende Punkte gibt, die keine Rechtskurve bilden, soll die Methode `false` zurückgeben.
* Testen Sie Ihre Methode, indem Sie die gegebenen 7 Punkte ins korrekte Format bringen, ihrer Methode übergeben, und den Rückgabewert ausgeben.

````java
class Turns {

    // Diese Methode müssen Sie nicht verändern
    public static boolean rightTurn(double x1, double y1, double x2, double y2, double x3, double y3) {
        if (((y2 - y1) * (x3 - x2)) - ((x2 - x1) * (y3 - y2)) > 0) {
            return true;
        }
        return false;
    }

    // Ihre Lösung
    public static boolean allRightTurns(double[][] points) {
        // Für jedes Tripel aufeinanderfolgender Punkte (i, i+1, i+2)
        for (int i = 0; i + 2 < points.length; i++) {
            double[] p1 = points[i];
            double[] p2 = points[i + 1];
            double[] p3 = points[i + 2];

            if (!rightTurn(p1[0], p1[1], p2[0], p2[1], p3[0], p3[1])) {
                return false;       // eine Nicht-Rechtskurve genügt
            }
        }
        return true;                // alle Tripel waren Rechtskurven
    }

    public static void main(String[] args) {
        double[] p1 = new double[]{0.0, 0.0};
        double[] p2 = new double[]{-0.84, 1.88};
        double[] p3 = new double[]{1.54, 2.7};
        double[] p4 = new double[]{2.18, 1.12};
        double[] p5 = new double[]{-1.44, 0.7};
        double[] p6 = new double[]{0.6, 2.0};
        double[] p7 = new double[]{0.8, -0.88};

        // Die 7 Punkte ins richtige Format bringen: ein Array von Punkten (double[][])
        double[][] points = {p1, p2, p3, p4, p5, p6, p7};

        System.out.println(allRightTurns(points));
    }
}
````

````java
Turns.main(new String[0]);
````

**Erklärung:** Ein Punkt ist ein `double[]` mit zwei Einträgen `{x, y}`, mehrere Punkte sind also ein `double[][]`. Wir gehen alle Tripel aufeinanderfolgender Punkte `(points[i], points[i+1], points[i+2])` durch und rufen die **gegebene** Methode `rightTurn` mit den passenden Koordinaten auf (`p[0]` ist x, `p[1]` ist y).

* Sobald ein Tripel **keine** Rechtskurve ist, können wir sofort `false` zurückgeben.
* Nur wenn die Schleife ohne Abbruch durchläuft, ist das Resultat `true`.
* Schleifengrenze: Das letzte Tripel beginnt bei `i = length - 3`, daher `i + 2 < points.length`. Sonst gibt es eine `ArrayIndexOutOfBoundsException`.
* Bei weniger als 3 Punkten gibt es keine Tripel, und die Methode gibt (leer erfüllt) `true` zurück.

**Hinweis:** Die Formel von `rightTurn` sollte man nicht nochmals selbst abtippen (wie in der Versuchsversion), sondern die Methode aufrufen. Das ist weniger fehleranfällig, und die Aufgabe sagt ausdrücklich, dass man sie verwenden soll.

Für die 7 gegebenen Punkte ist das Resultat `true`: Alle 5 Tripel `(p1,p2,p3)`, `(p2,p3,p4)`, `(p3,p4,p5)`, `(p4,p5,p6)` und `(p5,p6,p7)` sind Rechtskurven.
