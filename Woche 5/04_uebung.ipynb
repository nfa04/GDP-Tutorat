## Aufgabentyp 3: Fehlersuche

Finde und behebe die Fehler in den vier Programmen. Erkläre jeweils kurz, warum der ursprüngliche Code nicht kompiliert.

````java
class Aufgabe1 {

    public static void main(String[] args) {
        if (true) {
            int zahl = 10;
        }

        System.out.println(zahl);
    }
}

Aufgabe1.main(new String[0]);
````

````java
class Aufgabe2 {

    public static int berechne(int zahl) {
        if (zahl > 0) {
            int ergebnis = zahl * 2;
        } else {
            int ergebnis = zahl * 3;
        }

        return ergebnis;
    }

    public static void main(String[] args) {
        System.out.println(berechne(4));
    }
}

Aufgabe2.main(new String[0]);
````

````java
class Aufgabe3 {

    public static int quadriere(int zahl) {
        int ergebnis = zahl * zahl;
        return ergebnis;
    }

    public static void main(String[] args) {
        quadriere(4);
        System.out.println(ergebnis);
    }
}

Aufgabe3.main(new String[0]);
````

````java
class Aufgabe4 {

    public static int fakultaet(int n) {
        if (n <= 1) {
            int ergebnis = 1;
            return ergebnis;
        }

        ergebnis = n * fakultaet(n - 1);
        return ergebnis;
    }

    public static void main(String[] args) {
        System.out.println(fakultaet(4));
    }
}

Aufgabe4.main(new String[0]);
````
