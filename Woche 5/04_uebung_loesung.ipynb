Finde und behebe die Fehler in den vier Programmen. Erkläre jeweils kurz, warum der ursprüngliche Code nicht kompiliert.

````java
// Fehler: zahl war nur innerhalb des if-Blocks sichtbar.
class Aufgabe1 {
    public static void main(String[] args) {
        if (true) {
            int zahl = 10;
            System.out.println(zahl);
        }
    }
}

Aufgabe1.main(new String[0]);
````

## Aufgabentyp 4: Programmieren

**Aufgabe 4 – Rekursive Summe.** Ergänze `summeBis(int n)`: Für `n >= 0` liefert sie `1 + 2 + ... + n`. Verwende keine Schleife. Teste `0`, `1` und `4`.

````java
class RekursiveSumme {
    public static int summeBis(int n) {
        if (n == 0) {
            return 0;
        }
        return n + summeBis(n - 1);
    }

    public static void main(String[] args) {
        System.out.println(summeBis(0)); // 0
        System.out.println(summeBis(1)); // 1
        System.out.println(summeBis(4)); // 10
    }
}

RekursiveSumme.main(new String[0]);
````

## Aufgabentyp 6: Java-Abschlussaufgabe

**Gerade Ziffern zählen.** Ergänze die rekursive Methode `zaehleGeradeZiffern(int n)` für `n >= 0`. Sie zählt die geraden Dezimalziffern von `n`; auch `0` ist eine gerade Ziffer. Verwende weder Schleifen noch Strings. Teste: `0 → 1`, `135 → 0`, `2048 → 4`.

````java
class GeradeZiffern {
    public static int zaehleGeradeZiffern(int n) {
        int letzteZiffer = n % 10;
        int beitrag;
        if (letzteZiffer % 2 == 0) {
            beitrag = 1;
        } else {
            beitrag = 0;
        }

        if (n < 10) {
            return beitrag;
        }
        return beitrag + zaehleGeradeZiffern(n / 10);
    }

    public static void main(String[] args) {
        System.out.println(zaehleGeradeZiffern(0));    // 1
        System.out.println(zaehleGeradeZiffern(135));  // 0
        System.out.println(zaehleGeradeZiffern(2048)); // 4
    }
}

GeradeZiffern.main(new String[0]);
````

Bei `n < 10` ist nur noch eine Ziffer übrig. Sonst wird diese Ziffer gezählt und mit `n / 10` rekursiv weitergerechnet.
