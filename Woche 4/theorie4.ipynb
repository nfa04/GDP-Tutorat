## Aufgabentyp 3: "Fehlersuche"

Die folgenden drei Programme lassen sich nicht kompilieren. Finde heraus, warum der Java-Compiler meckert, und korrigiere den Code.

````java
// Programm 1: Gültigkeitsbereich (Scope)
class ScopeError {
    public static void main(String[] args) {
        int wert = 10;
        if (wert > 5) {
            int ergebnis = wert * 2;
        }
        System.out.println(ergebnis); 
    }
}

ScopeError.main(new String[0]);
````

````java
// Programm 2: Typenkonflikt
class TypError {
    public static int dividiere(int a, int b) {
        double resultat = (double) a / b;
        return resultat;
    }

    public static void main(String[] args) {
        System.out.println(dividiere(10, 3));
    }
}

TypError.main(new String[0]);
````

````java
// Programm 3: Fehlender Rückgabepfad
class MaxWert {
    public static int max(int a, int b) {
        if (a > b) {
            return a;
        } else if (b >= a) {
            return b;
        }
        // Hinweis: Weiss der Compiler, dass wir alle Fälle abgedeckt haben?
    }

    public static void main(String[] args) {
        System.out.println(max(7, 12));
    }
}

MaxWert.main(new String[0]);
````

## Programmfluss

Verfolge den Fluss des Programms. Was erscheint auf der Konsole?

````java
class Riddle {
    public static void main(String[] args){
        String str = "hallo";
        int i = Riddle.irgendwas(str);
        if(Riddle.nochmalwas(i) == "gut") {
            System.out.println("hallo");
        } else if(Riddle.wasanderes(str)) {
            System.out.println("du!");
        }
    }

    public static int irgendwas(String input) {
        return input.length();
    }

    public static String nochmalwas(int input) {
        if(input > 2) {
            return "gut";
        }
        return "schlecht";
    }

    public static boolean wasanderes(String input) {
        System.out.println(input);
        return false;
    }
}
````
