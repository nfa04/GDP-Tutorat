## Aufgabentyp 3: "Fehlersuche" (Lösungen)

````java
// Programm 1: Korrektur zu ScopeError.java
// Erklärung: 'ergebnis' muss ausserhalb des if-Blocks deklariert werden, damit es im main sichtbar ist.
class ScopeError {
    public static void main(String[] args) {
        int wert = 10;
        int ergebnis = 0;
        if (wert > 5) {
            ergebnis = wert * 2;
        }
        System.out.println(ergebnis); 
    }
}

ScopeError.main(new String[0]);
````

````java
// Programm 2: Korrektur zu TypError.java
// Erklärung: Der Rückgabetyp der Methode muss 'double' (oder float) sein, da 'resultat' ein Kommawert ist.
class TypError {
    public static double dividiere(int a, int b) {
        double resultat = (double) a / b;
        return resultat;
    }

    public static void main(String[] args) {
        System.out.println(dividiere(10, 3));
    }
}

TypError.main(new String[0]);
````

````java
// Programm 3: Korrektur zu MaxWert.java
// Erklärung: Das 'else if' muss durch ein einfaches 'else' ersetzt werden, damit garantiert immer ein Wert zurückgegeben wird.
class MaxWert {
    public static int max(int a, int b) {
        if (a > b) {
            return a;
        } else {
            return b;
        }
    }

    public static void main(String[] args) {
        System.out.println(max(7, 12));
    }
}

MaxWert.main(new String[0]);
````

**Programm 2 (IsCold):**
```java
public class WeatherUtils {
    public static boolean isCold(double temp) {
        if (temp < 15) {
            return true;
        } else {
            return false;
        }
    }

    public static void main(String[] args) {
        System.out.println(isCold(10));
    }
}
```
