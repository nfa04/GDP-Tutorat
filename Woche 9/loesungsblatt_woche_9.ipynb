## Teil 2: Fehlersuche in Java-Programmen – Musterlösung


### Aufgabe 2.1: Methodenaufruf

* **Fehlerbeschreibung:** Die Methode `addiere` in der Klasse `Rechner` ist als Instanzmethode (ohne `static`) deklariert. Im `main` wird sie jedoch direkt über den Klassennamen (`Rechner.addiere(3, 4)`) wie eine statische Methode aufgerufen.
* **Fehlerart:** **Kompilierfehler** (*non-static method addiere(int,int) cannot be referenced from a static context*).
* **Korrektur:** 
  - *Option A:* Instanz von `Rechner` erzeugen.
  - *Option B:* Methode `addiere` als `static` deklarieren.


```python
// Korrektur Option A (Instanz erzeugen):
class Rechner {
    int addiere(int a, int b) {
        return a + b;
    }
}

public class Main {
    public static void main(String[] args) {
        Rechner r = new Rechner();
        System.out.println(r.addiere(3, 4));
    }
}

// Alternativ Option B (static Schlüsselwort):
/*
class Rechner {
    static int addiere(int a, int b) {
        return a + b;
    }
}
*/
```

### Aufgabe 2.2: Der Konstruktor

* **Fehlerbeschreibung:** `public void Auto(String marke)` besitzt den Rückgabetyp `void`. Dadurch ist es kein Konstruktor, sondern eine normale Methode. Der Aufruf `new Auto("Toyota")` sucht nach einem Konstruktor mit String-Parameter, der nun nicht mehr existiert.
* **Fehlerart:** **Kompilierfehler** (*constructor Auto in class Auto cannot be applied to given types*).
* **Korrektur:** Den Rückgabetyp `void` entfernen.


```python
class Auto {
    String marke;

    // Korrekter Konstruktor ohne Rückgabetyp
    public Auto(String marke) {
        this.marke = marke;
    }
}

public class Main {
    public static void main(String[] args) {
        Auto a = new Auto("Toyota");
        System.out.println(a.marke); // Gibt "Toyota" aus
    }
}
```

### Aufgabe 2.3: Referenzen

* **Fehlerbeschreibung:** Bei `Person q = p;` verweist `p` noch auf `null`. Somit erhält auch `q` den Wert `null`. Die anschließende Zuweisung `p = new Person("Lena");` lässt `p` auf das neue Objekt zeigen, ändert aber die Referenz in `q` nicht (sie verweist weiterhin auf `null`). Der Zugriff `q.name` führt daher zu einer NullPointerException.
* **Fehlerart:** **Laufzeitfehler** (`NullPointerException`).
* **Korrektur:** Zuweisung `q = p;` erst nach der Instanziierung von `p` ausführen.


```python
class Person {
    String name;

    Person(String name) {
        this.name = name;
    }
}

public class Main {
    public static void main(String[] args) {
        Person p = new Person("Lena"); // Erst Objekt erzeugen
        Person q = p;                  // Jetzt zeigt q auf das Objekt "Lena"
        System.out.println(q.name);    // Gibt "Lena" aus
    }
}
```

### Aufgabe 2.4: Wertzuweisung im Konstruktor

* **Fehlerbeschreibung:** Die Zuweisungsrichtung im Konstruktor ist vertauscht: `betrag = kontostand;` weist dem Parameter `betrag` den uninitialisierten Standardwert des Feldes `kontostand` (`0.0`) zu. Das Attribut `kontostand` des Objekts bleibt unverändert `0.0`.
* **Fehlerart:** **Logikfehler / Semantischer Fehler** (Code kompiliert und läuft ohne Fehler, liefert aber ein falsches Ergebnis).
* **Korrektur:** Reihenfolge der Zuweisung umkehren: `kontostand = betrag;`.


```python
class Konto {
    double kontostand;

    Konto(double betrag) {
        this.kontostand = betrag; // Korrekte Zuweisung an das Feld
    }
}

public class Main {
    public static void main(String[] args) {
        Konto k = new Konto(250.0);
        System.out.println(k.kontostand); // Gibt 250.0 aus
    }
}
```

### Aufgabe 2.5: Statische Variablen (Static)

* **Fehlerbeschreibung:** `startnummer` ist als `static` deklariert. Eine statische Variable existiert nur ein einziges Mal für die gesamte Klasse. Wenn `s1` ("Mia") erzeugt wird, erhöht sich `startnummer` von 0 auf 1. Wenn `s2` ("Noah") erzeugt wird, erhöht sich dieselbe Variable von 1 auf 2. Beide Methodenaufrufe `s1.zeigeNummer()` und `s2.zeigeNummer()` geben daher `2` aus.
* **Fehlerart:** **Logikfehler / Designfehler**.
* **Korrektur:** Wenn jeder Spieler eine eigene, unveränderliche Startnummer behalten soll, benötigt man eine Instanzvariable `nummer` sowie einen statischen Zähler `naechsteNummer`.


```python
class Spieler {
    String name;
    static int naechsteNummer = 1; // Statischer Zähler für fortlaufende Nummern
    int nummer;                    // Instanzvariable für die individuelle Nummer

    Spieler(String name) {
        this.name = name;
        this.nummer = naechsteNummer++; // Speichert die aktuelle Nummer und erhöht den Zähler
    }

    void zeigeNummer() {
        System.out.println(this.name + " hat Startnummer: " + this.nummer);
    }
}

public class Main {
    public static void main(String[] args) {
        Spieler s1 = new Spieler("Mia");
        Spieler s2 = new Spieler("Noah");

        s1.zeigeNummer(); // Gibt 1 aus
        s2.zeigeNummer(); // Gibt 2 aus
    }
}
```

### Aufgabe 2.6: Parameter-Shadowing

* **Fehlerbeschreibung:** Namensüberdeckung (Shadowing). Die Parameter `breite` und `hoehe` im Konstruktor überdecken die gleichnamigen Instanzvariablen der Klasse. `breite = breite;` weist dem lokalen Parameter lediglich sich selbst zu. Die Instanzvariablen bleiben mit dem Standardwert `0.0` uninitialisiert.
* **Fehlerart:** **Logikfehler / Semantischer Fehler**.
* **Korrektur:** Explizite Instanzreferenz `this` verwenden (`this.breite = breite;`).


```python
class Rechteck {
    double breite;
    double hoehe;

    Rechteck(double breite, double hoehe) {
        this.breite = breite; // this verhindert Namensüberdeckung
        this.hoehe = hoehe;
    }

    double berechneFlaeche() {
        return breite * hoehe;
    }
}

public class Main {
    public static void main(String[] args) {
        Rechteck r = new Rechteck(5.0, 3.0);
        System.out.println(r.berechneFlaeche()); // Gibt 15.0 aus
    }
}
```

### Aufgabe 1a: Übersetzung von Programm 1 (Funktion & Main) in Java

```python
public class Main {
    public static double test(double a) {
        double b = a + a;
        System.out.println((int) b); // Gibt 8 bzw. 18 als Ganzzahl aus
        double c = Math.sqrt(a);
        return c;
    }

    public static void main(String[] args) {
        int a = 4;
        int b = 9;
        
        // In Python 'False' / 'True', in Java klein geschrieben: 'false' / 'true'
        System.out.println(a == b); // false
        System.out.println(a != b); // true
        
        double c = test(a); // Aufruf für a=4 -> erzeugt 8 und liefert 2.0
        System.out.println(c);
        
        c = test(b);        // Aufruf für b=9 -> erzeugt 18 und liefert 3.0
        System.out.println(c);
    }
}
```

### Aufgabe 1b: Übersetzung von Programm 2 (Klasse Punkt & Main) in Java

```python
class Punkt {
    // Statische Klassenvariable (entspricht Punkt.ursprung in Python)
    static Punkt ursprung = new Punkt(0, 0);

    // Instanzvariablen (Objektdaten)
    double x;
    double y;

    // Konstruktoren (Überladung zur Unterstützung des optionalen y-Parameters aus Python)
    Punkt(double x) {
        this(x, 0.0);
    }

    Punkt(double x, double y) {
        this.x = x;
        this.y = y;
    }

    // Statische Klassenmethode (entspricht @staticmethod in Python)
    public static void hallo(String name) {
        System.out.println("Hallo " + name);
    }

    // Objekt- / Instanzmethoden
    public void setzeY(double y) {
        this.y = y;
    }

    public double abstand() {
        double xabs = this.x - ursprung.x;
        double yabs = this.y - ursprung.y;
        return Math.sqrt(xabs * xabs + yabs * yabs);
    }
}

public class Main {
    public static void main(String[] args) {
        String name = "Fritz";
        Punkt.hallo(name);

        Punkt p = new Punkt(3);
        System.out.println(p.abstand()); // 3.0

        p.setzeY(4);
        System.out.println(p.abstand()); // 5.0

        Punkt.ursprung.x = 3;
        Punkt.ursprung.y = 4;
        System.out.println(p.abstand()); // 0.0
    }
}
```

### Aufgabe 2: Implementierung der 3D-Vektorklasse in Python

```python
import math

class Vektor3D:
    def __init__(self, x: float, y: float, z: float):
        self.x = float(x)
        self.y = float(y)
        self.z = float(z)

    # Objektmethode: Berechnet die euklidische Länge des Vektors
    def laenge(self) -> float:
        return math.sqrt(self.x * self.x + self.y * self.y + self.z * self.z)

    # Statische/Klassenmethode: Addiert zwei Vektoren und gibt einen neuen Vektor zurück
    @staticmethod
    def addiere(v1: 'Vektor3D', v2: 'Vektor3D') -> 'Vektor3D':
        return Vektor3D(v1.x + v2.x, v1.y + v2.y, v1.z + v2.z)

    # Statische/Klassenmethode: Skalarmultiplikation (gibt neuen Vektor zurück)
    @staticmethod
    def skalarmult(v: 'Vektor3D', s: float) -> 'Vektor3D':
        return Vektor3D(v.x * s, v.y * s, v.z * s)

    def __str__(self) -> str:
        return f"Vektor3D({self.x}, {self.y}, {self.z})"


# Test und Demonstration aller geforderten Methoden
if __name__ == "__main__":
    v1 = Vektor3D(1.0, 2.0, 2.0)
    v2 = Vektor3D(4.0, 0.0, 3.0)

    # 1. Objektmethode: laenge()
    print("v1:", v1)
    print("Länge von v1:", v1.laenge()) # sqrt(1^2 + 2^2 + 2^2) = 3.0

    # 2. Klassenmethode: addiere()
    v3 = Vektor3D.addiere(v1, v2)
    print("v3 (v1 + v2):", v3)          # Vektor3D(5.0, 2.0, 5.0)

    # 3. Klassenmethode: skalarmult()
    v4 = Vektor3D.skalarmult(v1, 3.0)
    print("v4 (3 * v1):", v4)           # Vektor3D(3.0, 6.0, 6.0)
    print("Länge von v4:", v4.laenge()) # 9.0
```
