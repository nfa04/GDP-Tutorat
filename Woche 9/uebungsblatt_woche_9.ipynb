### Aufgabe 2.5: Statische Variablen (Static)

````python
class Spieler {
    String name;
    static int startnummer = 0;

    Spieler(String name) {
        this.name = name;
        startnummer++;
    }

    void zeigeNummer() {
        System.out.println(this.startnummer);
    }
}

public class Main {
    public static void main(String[] args) {
        Spieler s1 = new Spieler("Mia");
        Spieler s2 = new Spieler("Noah");

        s1.zeigeNummer();
        s2.zeigeNummer();
    }
}
````

### Aufgabe 2.6: Parameter-Shadowing

````python
class Rechteck {
    double breite;
    double hoehe;

    Rechteck(double breite, double hoehe) {
        breite = breite;
        hoehe = hoehe;
    }

    double berechneFlaeche() {
        return breite * hoehe;
    }
}

public class Main {
    public static void main(String[] args) {
        Rechteck r = new Rechteck(5.0, 3.0);
        System.out.println(r.berechneFlaeche());
    }
}
````

### Programm 2

##### Python-Code:
```python
import math

class Punkt:
    ursprung = None

    def __init__(self, x, y=0.0):
        self.x = x
        self.y = y

    @staticmethod
    def hallo(name):
        print("Hallo " + name)

    def setze_y(self, y):
        self.y = y

    def abstand(self):
        xabs = self.x - Punkt.ursprung.x
        yabs = self.y - Punkt.ursprung.y
        return math.sqrt(xabs * xabs + yabs * yabs)

# Ursprung initialisieren
Punkt.ursprung = Punkt(0, 0)

if __name__ == "__main__":
    name = "Fritz"
    Punkt.hallo(name)
    
    p = Punkt(3)
    print(p.abstand())
    
    p.setze_y(4)
    print(p.abstand())
    
    Punkt.ursprung.x = 3
    Punkt.ursprung.y = 4
    print(p.abstand())
```

##### Output:
```text
Hallo Fritz
3.0
5.0
0.0
```
